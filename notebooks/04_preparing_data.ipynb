```python
import pandas as pd
from pathlib import Path
```

```python
FEATURED_DIR = Path('../data/featured/')
FEATURED_DATA = Path.joinpath(FEATURED_DIR , 'featured_dataset-electric.csv')
TARGET_Y = 'PJME_MW'
```

```python
# Create split directory
SPLIT_DIR = Path('../data/split/')
SPLIT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
featured_df = pd.read_csv(FEATURED_DATA)
```

# **Define reusable functions**

```python
def print_line(num):
    print('=' * num)
    
def print_txt(text, num=50):
    print_line(num)
    print(text)
    print_line(num)
```

```python
featured_df['Datetime'] = pd.to_datetime(featured_df['Datetime'])
```

```python
print_txt('Define target and features')

# Define target and features
X = featured_df.drop(columns=['Datetime', 'PJME_MW'])
y = featured_df[TARGET_Y]
print('features     :')
print(*X.columns)
print_line(50)
print('taget column :')
print(TARGET_Y)
```

```python
# Chronological 80/20 split
split_index = int(len(featured_df) * 0.8)
```

```python
print_txt('Combine features, target, and Datetime')
train_df = featured_df.iloc[:split_index].copy()
test_df = featured_df.iloc[split_index:].copy()
print(f'Training set  : {train_df.shape}')
print(f'Testing set   : {test_df.shape}')
```

```python
print("\nTraining period :")
print(featured_df['Datetime'].iloc[0])
print(featured_df['Datetime'].iloc[split_index - 1])

print("\nTesting period  :")
print(featured_df['Datetime'].iloc[split_index])
print(featured_df['Datetime'].iloc[-1])
```

```python
print_txt('Export datasets')

train_df.to_csv(SPLIT_DIR / 'train.csv', index=False)
test_df.to_csv(SPLIT_DIR / 'test.csv', index=False)

print("Training and testing datasets exported successfully!")
```
